```python
import os
import json
import matplotlib.pyplot as plt
import numpy as np

# 配置样式
plt.rcParams['figure.figsize'] = (22, 15)
plt.rcParams['xtick.major.width'] = 2
plt.rcParams['ytick.major.width'] = 2
plt.rcParams['axes.grid'] = False
plt.rcParams['axes.linewidth'] = 2
plt.rcParams['ytick.major.size'] = 12
plt.rcParams['xtick.major.size'] = 12
plt.rcParams['axes.titlesize'] = 52
plt.rcParams['axes.labelsize'] = 52
plt.rcParams['lines.linewidth'] = 6
plt.rcParams['lines.markersize'] = 24
plt.rcParams['xtick.labelsize'] = 52
plt.rcParams['ytick.labelsize'] = 52
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['DejaVu Serif']
plt.rcParams['legend.fontsize'] = 42
plt.rcParams['figure.dpi'] = 600

# 数据目录
data_dir = "./"
files = [
    "gemma27b-lmsys-distribution.jsonl",
    "gemma27b-sharegpt-distribution.jsonl",
    "llama8b-lmsys-distribution.jsonl",
    "llama8b-sharegpt-distribution.jsonl",
]

def format_title(filename):
    model_part, dataset_part, other = filename.replace('.jsonl', '').split('-')
    model_name = model_part.replace('gemma27b', 'Gemma-27B').replace('llama8b', 'LLaMA-8B')
    dataset_name = dataset_part.strip('-').replace('lmsys', 'LMSYS').replace('sharegpt', 'ShareGPT')
    return f"{model_name} {dataset_name}"

def count_in_range(output_lens, min_val=0, max_val=820):
    return np.sum((output_lens >= min_val) & (output_lens <= max_val)) / len(output_lens)

# 绘图
for filename in files:
    path = os.path.join(data_dir, filename)
    input_lens, output_lens = [], []
    with open(path, 'r') as f:
        for line in f:
            obj = json.loads(line)
            input_lens.append(obj['input_length'])
            output_lens.append(obj['generated_length'])

    input_lens = np.array(input_lens)
    output_lens = np.array(output_lens)
    
    # Print count of outputs in range 0-820
    count = count_in_range(output_lens)
    print(f"{format_title(filename)}: {count} samples have generated length in range 0-820")

    plt.figure()
    plt.hist(input_lens,  bins=100, alpha=0.6, density=True, color='#4d5de7', label=f"Input (mean: {input_lens.mean():.0f})")
    plt.hist(output_lens, bins=100, alpha=0.6, density=True, color='#a45753', label=f"Output (mean: {output_lens.mean():.0f})")
    plt.title(format_title(filename))
    plt.xlabel("# Tokens")
    plt.ylabel("Density")
    plt.legend()
    plt.tight_layout()
    # plt.show()
    plt.savefig(fname=filename.replace('.jsonl', '.pdf'), dpi=600, format="pdf")
    plt.close()
```
