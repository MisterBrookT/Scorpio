```python
import matplotlib.pyplot as plt
import matplotlib.lines as mlines

# 调整后的顺序：让 Scorpio(Ours) 放在最前面
labels = ['Scorpio(Ours)', 'vLLM', 'Mooncake', 'S3']
colors = ['#ec6446', '#f29d46', '#3f906e', '#836ed5']     # 对应新顺序
markers = ['s', '^', 'D', 'o']
linestyles = ['-', '--', '--', '-']
linewidth = 6
legend_fontsize = 52  # 可根据实际需求调整

# 创建图例图形
fig, ax = plt.subplots(figsize=(25, 1))  # 宽度根据label数调整
handles = [
    mlines.Line2D(
        [], [], color=colors[i], marker=markers[i], linestyle=linestyles[i],
        linewidth=linewidth, label=label, markersize=32
    )
    for i, label in enumerate(labels)
]

# 添加图例并保存
ax.legend(handles=handles, loc='center', fontsize=legend_fontsize, ncol=len(labels), frameon=False)
ax.axis('off')  # 去除坐标轴
plt.savefig("slo_goodput_legend.pdf", dpi=600, format="pdf", bbox_inches="tight")
plt.close()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.lines as mlines

# 设置样式
legend_fontsize = 52
linewidth = 4
marker_size = 18

# 创建图例图形
fig, ax = plt.subplots(figsize=(20, 1))  # 宽度适配三个元素
ax.axis('off')  # 不显示坐标轴

# 三个图例项
ttft_patch = mpatches.Patch(color="#ec6446", label="TTFT Violation", edgecolor='black', linewidth=1.5)
tpot_patch = mpatches.Patch(color="#836ed5", label="TPOT Violation", edgecolor='black', linewidth=1.5)
goodput_line = mlines.Line2D(
    [], [], color='black', linestyle='--',
    marker='o', markersize=marker_size,
    markerfacecolor='white', markeredgecolor='black', markeredgewidth=2.5,
    linewidth=linewidth, label='Goodput'
)

# 绘制图例
handles = [ttft_patch, tpot_patch, goodput_line]
ax.legend(
    handles=handles,
    loc='center',
    fontsize=legend_fontsize,
    ncol=len(handles),
    frameon=False
)

# 保存图例
plt.savefig("ablation_legend.pdf", dpi=600, format="pdf", bbox_inches="tight")
plt.close()
```
